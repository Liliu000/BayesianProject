```python

def split_residence(row):
    parts = [p.strip() for p in row.split(',')]
    
    if len(parts) == 3:
        return parts[0], parts[2] # City, Country
    elif len(parts) == 2:
        return parts[0], parts[1] # City, Country
    else:
        return parts[0], None      # City, Missing Country

# Apply the split
df2014[['City', 'Country']] = df2014['residence'].apply(lambda x: pd.Series(split_residence(x)))
df2014['City'] = df2014['City'].str.strip()
df2014['Country'] = df2014['Country'].str.strip()
print(df2014.isna().sum())

df2014.head()
```

```python
unique_mapping = df_country_city.drop_duplicates('City', keep=False)
mapping_dict = unique_mapping.set_index('City')['Country'].to_dict()

# 2. Get the names of cities that appear with multiple countries
# This uses the 'duplicates' logic you already found
ambiguous_city_names = df_country_city[df_country_city.duplicated('City', keep=False)]['City'].unique()

# Create a mask for rows with missing countries
missing_mask = df2014['Country'].isna()

# 3. Separate rows where the city exists but has multiple possible countries
df_ambiguous = df2014[missing_mask & df2014['City'].isin(ambiguous_city_names)].copy()

# (Optional) Remove these from the main dataframe so you don't process them twice
# df2014 = df2014[~(missing_mask & df2014['City'].isin(ambiguous_city_names))]

# 4. Map the unique countries to the missing values
df2014.loc[missing_mask, 'Country'] = df2014.loc[missing_mask, 'Country'].fillna(df2014['City'].map(mapping_dict))

print(f"Moved {len(df_ambiguous)} ambiguous rows to df_ambiguous.")
print(f"Filled remaining missing countries using unique mappings.")

print(df2014.isna().sum())
print(len(df_ambiguous['City'].unique()), "ambiguous city names:", df_ambiguous['City'].unique())
df_ambiguous.head()
```

# Weather


```python
import openpyxl
compass_to_deg = {
    'N': 0, 'NNE': 22.5, 'NE': 45, 'ENE': 67.5, 'E': 90, 'ESE': 112.5, 'SE': 135, 'SSE': 157.5,
    'S': 180, 'SSW': 202.5, 'SW': 225, 'WSW': 247.5, 'W': 270, 'WNW': 292.5, 'NW': 315, 'NNW': 337.5,
    'VAR': np.nan, 'CALM': np.nan # Variable or calm winds don't have a distinct direction
}

def clean_weather_data(file_path, sheet_name):
    # 1. Load the Excel file
    df_weather = pd.read_excel(file_path, sheet_name=sheet_name)
    cols_to_clean = ['Temperature', 'Dew Point','Wind Speed','Precip.']

    # 2. Extract the numbers using pandas string extraction
    for col in cols_to_clean:
        # r'(-?\d+\.?\d*)' looks for optional minus signs, digits, and optional decimals
        # [0] grabs the first match, and .astype(float) turns it into a mathematical number
        df_weather[col] = df_weather[col].str.extract(r'(-?\d+\.?\d*)')[0].astype(float)
    
    df_weather['Wind_Deg'] = df_weather['Wind'].map(compass_to_deg)
    df_weather['Wind_Deg'] = df_weather['Wind_Deg'].fillna(0)

    speed_for_calc = df_weather['Wind Speed'].where(df_weather['Wind'].isin(compass_to_deg.keys()) & (df_weather['Wind'] != 'VAR'), 0)

    # 4. Calculate the Tailwind Component using Trigonometry
    # The "ideal" tailwind comes from WSW (247.5 degrees)
    df_weather['Tailwind_Component'] = speed_for_calc * np.cos(np.radians(df_weather['Wind_Deg'] - 247.5))

    cols_to_save = ['Temperature', 'Dew Point','Precip.', 'Tailwind_Component']
    mean_weather = df_weather[cols_to_save].mean()
    return mean_weather

results = []

for year in [2014, 2015, 2016, 2017, 2018, 2019]:
    df_weather_bos = clean_weather_data("Book1.xlsx", sheet_name=f"{year}-Bos")
    df_weather_hop = clean_weather_data("Book1.xlsx", sheet_name=f"{year}-Hop")
    combined_avg = (df_weather_bos + df_weather_hop) / 2
    
    # 3. Store the result in a dictionary
    row = {
        'Year': year, 
        'Temperature': combined_avg['Temperature'], 
        'Dew Point': combined_avg['Dew Point'], 
        'Precip.': combined_avg['Precip.'], 
        'Tailwind_Component': combined_avg['Tailwind_Component']
    }
    results.append(row)

df_weather_avg = pd.DataFrame(results)
print(df_weather_avg)

df_weather_avg.to_csv("data/weather_averages.csv", index=False)
#df2014w = {'Temperature': (df2014_wb['Temperature'], df2014_wh['Temperature']), 'Humidity': (df2014_wb['Humidity'], df2014_wh['Humidity'])}
```
