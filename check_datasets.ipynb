```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

```

```python
df = pd.read_csv("data/filtered_marathons_results.csv")
print(df.isna().sum())
print(df.shape)
print(df.columns)
```

```python
all_repeats = df[df.duplicated(subset=['Name','M/F', 'City'], keep=False)].sort_values(['Name', 'City'])
print(all_repeats.shape)


#all_repeats.to_csv("data/repeated_names.csv", index=False)
```

```python
counts = df.groupby(['Name', 'M/F', 'City'])['Name'].transform('size')
more_than_3 = df[counts > 2].sort_values(['Name', 'M/F', 'City'])

print(f"Found {more_than_3['Name'].nunique()} names repeated more than 3 times.")
df_temp = all_repeats.groupby(['Name', 'M/F', 'City']).size().reset_index(name='Count')
print(df_temp.head())
df_temp.to_csv("data/name_counts.csv", index=False)
```

# Find overall repeated names

```python
df_temp = dfall_overall.groupby(['Name', 'M/F', 'City']).size().reset_index(name='Count')

def print_repeated_names(df_temp, threshold):
    more_than_threshold = df_temp[df_temp['Count'] > threshold].sort_values(['Name', 'M/F', 'City'])
    print(f"Found {more_than_threshold['Name'].nunique()} names repeated more than {threshold} times.")
    return more_than_threshold

more_than_3 = print_repeated_names(df_temp, 3)
more_than_4 = print_repeated_names(df_temp, 4)
more_than_5 = print_repeated_names(df_temp, 5)
```
